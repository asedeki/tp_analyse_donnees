## Section 9 — Les boucles `for`

### À quoi ça sert ?

Une **boucle** répète une action pour chaque élément.

**Règles :** `:` obligatoire, code indenté.

### Exemple

````python
mesures_boucle = [20.1, 21.3, 19.8, 22.0, 20.5]
for m in mesures_boucle:
    print(m)

somme_boucle = 0
for m in mesures_boucle:
    somme_boucle = somme_boucle + m
print("Somme :", somme_boucle)

compte_boucle = 0
for m in mesures_boucle:
    if m > 20:
        compte_boucle = compte_boucle + 1
print("Au-dessus de 20 :", compte_boucle)
````

## Section 14 — Ajuster une gaussienne avec SciPy

### Le problème

On a des mesures dispersées. On veut : (1) autour de quelle valeur elles se groupent, (2) à quel point elles sont étalées.

### Étape 1 — Voir les données

````python
import numpy as np
import matplotlib.pyplot as plt
rng_gauss = np.random.default_rng(3)
longueurs_exemple = 150.0 + rng_gauss.normal(0, 1.5, 50)
plt.hist(longueurs_exemple, bins=10, edgecolor="k")
plt.xlabel("Longueur (mm)")
plt.ylabel("Nombre")
plt.show()
````

### Étape 2 — Ajuster avec SciPy

`norm.fit` renvoie `mu` (centre) et `sigma` (largeur).

````python
from scipy.stats import norm
mu_exemple, sigma_exemple = norm.fit(longueurs_exemple)
print(f"mu = {mu_exemple:.2f} mm")
print(f"sigma = {sigma_exemple:.2f} mm")
````

### Étape 3 — Courbe théorique

`np.linspace` fabrique une grille, `norm.pdf` la hauteur.

````python
x_exemple_gauss = np.linspace(longueurs_exemple.min() - 3, longueurs_exemple.max() + 3, 200)
y_exemple_gauss = norm.pdf(x_exemple_gauss, mu_exemple, sigma_exemple)
````

### Étape 4 — Superposer

````python
plt.hist(longueurs_exemple, bins=10, density=True, edgecolor="k", alpha=0.6, label="donnees")
plt.plot(x_exemple_gauss, y_exemple_gauss, "r-", label=f"mu={mu_exemple:.2f}, sigma={sigma_exemple:.2f}")
plt.xlabel("Longueur (mm)")
plt.ylabel("Densite")
plt.legend()
plt.show()
````
